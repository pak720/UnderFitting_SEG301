```python
# XÓA DUPLICATES

import pandas as pd

# Đọc dữ liệu
df = pd.read_csv("Data_infodoanhnghiep.csv", encoding="utf-8-sig")

# Xóa dòng trùng theo các cột index 0, 2, 3 (giữ dòng đầu tiên)
df = df.drop_duplicates(subset=df.columns[[0, 2, 3]], keep="first")

# Reset index (khuyến nghị)
df.reset_index(drop=True, inplace=True)

# Ghi đè lại vào file cũ
df.to_csv("Data_infodoanhnghiep_clean.csv", index=False, encoding="utf-8-sig")

# Kiểm tra nhanh
df.info()

```

```python
# CHUYỂN CSV SANG JSON

import csv
import json

def csv_to_json(csv_path, json_path):
    data = []

    # utf-8-sig để loại bỏ BOM
    with open(csv_path, mode='r', encoding='utf-8-sig', newline='') as csv_file:
        reader = csv.DictReader(csv_file)
        for row in reader:
            data.append(row)

    with open(json_path, mode='w', encoding='utf-8') as json_file:
        json.dump(data, json_file, ensure_ascii=False, indent=4)

if __name__ == "__main__":
    csv_file_path = "Data_infodoanhnghiep_clean.csv"
    json_file_path = "Data_infodoanhnghiep.json"
    csv_to_json(csv_file_path, json_file_path)
```
